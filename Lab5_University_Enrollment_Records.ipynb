```python
LAST_NAME = "Manansala"
STUDENT_ID = "TUPM-26-4359"
FAVORITE_ARTIST = "Joshua Garcia"
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

department_a = sorted([ID_SUM + (i * SEED_DIGIT) for i in range(5)])
department_b = sorted([NAME_LENGTH + (i * ARTIST_LENGTH) for i in range(5)])

print("Department A dataset generated:", department_a)
print("Department B dataset generated:", department_b)
```

```python
def consolidate_datasets(list_a, list_b):
    merged = []
    i = 0
    j = 0
    comparisons = 0

    while i < len(list_a) and j < len(list_b):
        comparisons += 1
        if list_a[i] <= list_b[j]:
            merged.append(list_a[i])
            i += 1
        else:
            merged.append(list_b[j])
            j += 1

    while i < len(list_a):
        merged.append(list_a[i])
        i += 1
    
    while j < len(list_b):
        merged.append(list_b[j])
        j += 1
    
    return merged, comparisons

consolidated_dataset, comparison_count = consolidate_datasets(department_a, department_b)
```

```python
records_before = len(department_a) + len(department_b)
records_after = len(consolidated_dataset)
is_ordered = all(consolidated_dataset[k] <= consolidated_dataset[k + 1] for k in range(len(consolidated_dataset) -1))
```
